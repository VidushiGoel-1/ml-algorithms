```python
from sklearn.datasets import make_regression
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

# 1. Create data: 2 features -> 1 output (so we can draw a 3D plane)


```python
X, y = make_regression(n_samples=100, n_features=2, n_informative=2,
                       n_targets=1, noise=50, random_state=1)
df = pd.DataFrame({"cgpa": X[:, 0], "iq": X[:, 1], "lpa": y})
print(df.head())
```

# 2. 3D view of the raw data


```python
fig = px.scatter_3d(df, x="cgpa", y="iq", z="lpa")
fig.show()
```

# 3. Split


```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=3)
```

# 4. Train


```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

# 5. Inspect the learned weights: lpa = b0 + b1*cgpa + b2*iq


```python
print("Intercept (b0):", lr.intercept_)
print("Coefficients (b1, b2):", lr.coef_)
```

# 6. Predict and evaluate


```python
y_pred = lr.predict(X_test)
print("MAE :", mean_absolute_error(y_test, y_pred))
print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2  :", r2_score(y_test, y_pred))
```

# 7. Draw the fitted plane (the hyperplane from your diagram)


```python
x1 = np.linspace(X[:, 0].min(), X[:, 0].max(), 20)
x2 = np.linspace(X[:, 1].min(), X[:, 1].max(), 20)
xGrid, yGrid = np.meshgrid(x1, x2)
final = lr.intercept_ + lr.coef_[0] * xGrid + lr.coef_[1] * yGrid

fig = go.Figure(data=[
    go.Scatter3d(x=df["cgpa"], y=df["iq"], z=df["lpa"], mode="markers"),
    go.Surface(x=x1, y=x2, z=final, opacity=0.5),
])
fig.update_layout(scene=dict(xaxis_title="cgpa", yaxis_title="iq", zaxis_title="lpa"))
fig.show()
```
